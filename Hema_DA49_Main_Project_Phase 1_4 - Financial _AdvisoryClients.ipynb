<h4 style="margin:0 0 8px 0;color:#0f172a;">Missing-Value Identification</h4>

```python
miss_v=df.isnull().sum()
missing_values=miss_v[miss_v>0]
print(missing_values)
```

 <h4 style="margin:0 0 8px 0;color:#0f172a;">Missing-Value Filling Using Immutation </h3>

```python
mean_cols=["credit_score","trust_in_institutions_1_10","user_satisfaction_1_5"]
df[mean_cols]=df[mean_cols].fillna(df[mean_cols].mean())
df["monthly_expenses_usd"]=df["monthly_expenses_usd"].fillna(df["monthly_expenses_usd"].median())
df["annual_income_usd"]=df["annual_income_usd"].fillna(df["annual_income_usd"].median())
df["investment_portfolio_usd"]=df["investment_portfolio_usd"].fillna(df["investment_portfolio_usd"].median())

df['log_income']=df['log_income'].fillna(np.log(df['annual_income_usd']))
#the income_x_debt missing values filled using annual_income_usd * total_debt_usd
df['income_x_debt']=df['income_x_debt'].fillna(df['annual_income_usd']*df['total_debt_usd'])
print("Number of columns having missing values:")
df.isnull().sum()
```

    <h2 style="margin:0;color:white;font-size:25px;">5. Outlier Detection Using - IQR Method</h2>
    <p style="margin:5px 0 0 0;color:#dbeafe;font-size:14px;">Identify unusual high or low numerical observations without automatically deleting them</p>
</div>

```python
df.describe()
```

```python
outlier_cols = [
    'Emergency_Fund_Months',
    'Investment_Portfolio_Usd',
    'Time_Horizon_Years',
    'Risk_Tolerance_1_10',
    'Session_Duration_Min',
    'User_Satisfaction_1_5'
]
for col in outlier_cols:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)

    IQR=Q3-Q1

    lower_limit=Q1-1.5*IQR
    upper_limit=Q3+1.5*IQR

    outliers=df[(df[col]<lower_limit) | (df[col]>upper_limit)]

    print(col,":",len(outliers),"outliers")
    #print(outliers[[col]].head(10))
```

<table style="width:100%;border-collapse:collapse;margin-top:14px;">
<tr style="background:#1d4ed8;color:white;">
<th style="padding:10px;text-align:left;">Enginerred Feature</th><th style="padding:10px;text-align:left;">Purpose</th>
</tr>
<tr><td style="padding:9px;border:1px solid #d1fae5;">Monthly_Income_Usd</td><td style="padding:9px;border:1px solid #d1fae5;">Created column for monthly income from annual income column</td></tr>
<tr><td style="padding:9px;border:1px solid #d1fae5;">Disposable_Income_Usd</td><td style="padding:9px;border:1px solid #d1fae5;">Column to display howmuch income remains after expenses monthly</td></tr>
<tr><td style="padding:9px;border:1px solid #d1fae5;">Emergency_Fund_Amount</td><td style="padding:9px;border:1px solid #d1fae5;">Empergency fund amount calculated by using monthly expenses and emergency fund months</td></tr>
<tr><td style="padding:9px;border:1px solid #d1fae5;">Age_Group</td><td style="padding:9px;border:1px solid #d1fae5;">Groups clients into age-based categories</td></tr>
<tr><td style="padding:9px;border:1px solid #d1fae5;">Risk_Category</td><td style="padding:9px;border:1px solid #d1fae5;">Groups clients risk levels into risk-based categories</td></tr>
<tr><td style="padding:9px;border:1px solid #d1fae5;">Total_NumberOfTopics_Discussed</td><td style="padding:9px;border:1px solid #d1fae5;">Count of topics discussed by each client</td></tr>
</table>

```python
# Monthly expenses already present, so monthly income also created from annual income
df["Monthly_Income_Usd"]=(df["Annual_Income_Usd"]/12).round(2)

# Calculate howmuch income remains after expenses monthly
df['Disposable_Income_Usd'] = (df['Monthly_Income_Usd'] -df['Monthly_Expenses_Usd']).round(2)

# Emergency fund month already present,so empergency fund per month calculated by using monthly expenses and emergency fund months
df["Emergency_Fund_Amount"]= (df["Monthly_Expenses_Usd"]*df["Emergency_Fund_Months"]).round(2)

# Age group
df['Age_Group']=pd.cut(df['Age'],bins=[0,30,45,60,100],labels=['Young','Middle-Aged','Senior','Older'])

# Risk category
df['Risk_Category'] = pd.cut(df['Risk_Tolerance_1_10'],bins=[0, 3, 7, 10],labels=['Low Risk', 'Moderate Risk', 'High Risk'])

# Column to get the number of topics discussed by each client
df["Total_NumberOfTopics_Discussed"]=df['Topics_Discussed'].str.split(";").str.len()

df.sample(10)
```

### 1. Client Demographic Profile

The categorical analysis shows that Older clients form the largest age group, with 1,352 clients, followed by Senior clients with 1,213, Young clients with 1,266 and Middle-Aged clients with 1169.

The average client age is approximately 46.8 years, with a median of 46 years. Ages range from 18 to 85, and the middle 50% of clients fall between 30 and 61 years. The age histogram confirms a broad, multi-peaked distribution rather than concentration in a single life stage.
This indicates that the advisory dataset represents clients across different stages of life and therefore requires more than a single age-based advisory strategy.

### 2. Employment Status and Financial Position
Salaried clients are the largest group, with 1,976 clients, followed by Retired, Self-Employed, Student, and Unemployed clients. Salaried and Retired clients make up about 60% of the client base.

Self-Employed clients have the highest average income (174,437 usd) and debt (274,187 usd), followed by Salaried clients with average income of 149,407 usd and 234,527 usd. Students and Unemployed clients have the lowest income, debt, and monthly expenses. Retired, student and unemployed clients have substantially lower average debt.

Overall, employment status is strongly related to clients' financial capacity, debt, and spending.

### 3. Income, Debt, Expenses and Disposable Income

Annual Income, debt, expenses, and disposable income are strongly right-skewed, with most clients in lower-to-moderate ranges and a few with very high values.

Income ranges from approximately 1,000 usd to more than 1.34 million usd whereas annual income has a mean of approximately 108,688 usd, while the median is 90,413 usd, showing the effect of high-income clients.

Income and debt have a positive relationship: higher-income clients generally tend to have higher debt and the skewness analysis further confirms that annual income, investment portfolio, income and debt contain substantial high-value observations. Investment portfolio has the highest skewness and peakedness, while income and debt also demonstrate strong positive skewness.

Overall, the median is more representative than the mean for a typical client.

### 4. Investment Portfolio Behaviour

There is a positive relationship between age and investment portfolio size, with a correlation of about 0.51. Older clients generally hold larger investment portfolios, which is consistent with longer periods of wealth accumulation, although age alone does not completely explain portfolio size.

Higher-income clients also tend to have larger investment portfolios, although there is some variation.

Older and Senior clients generally have larger portfolios, while High Risk clients tend to have smaller portfolios. Moderate and Low Risk clients show greater variation and contain the highest-value portfolio observations. The top five portfolio holders have investments of about 3.29–3.81 million usd and are aged 58–63.

Overall, age and income are important factors associated with larger investment portfolios.

### 5. Risk Profile

The risk-category analysis shows that Moderate Risk clients dominate the dataset, representing approximately 78% of clients. Low Risk clients account for approximately 21%, while High Risk clients represent only around 0.6%.
Risk tolerance varies considerably by age group. Young clients have the highest average risk tolerance at approximately 5.39, whereas Middle-Aged clients have the lowest at approximately 2.72. Older clients average approximately 4.38 and Senior clients approximately 3.51.

High Risk clients have higher risk-tolerance scores, with Middle-Aged High Risk clients showing the highest average (8.66). However, this is expected because Risk Category is derived from the risk-tolerance score, and the High Risk group is very small.


Marital status, in contrast, has little distinguishing effect on risk tolerance. Even though the violin plots for marital-status groups have very similar distributions and median positions,The Married group has the highest number of observations in all risk categories, mainly because it represents the largest group in the dataset. High Risk cases are very few across all marital-status groups, with no High Risk cases observed among Widowed individuals.

### 6. Emergency-Fund Preparedness

Emergency-fund coverage differs across risk categories.

Moderate Risk: approximately 3.4 months
Low Risk: approximately 3.3 months
High Risk: approximately 2.5 months

High Risk clients therefore appear to have weaker emergency financial buffers. The wider confidence interval for the High Risk group also indicates greater variability or uncertainty in emergency-fund coverage.

This is an important practical finding because emergency-fund preparedness should be considered alongside investment risk rather than treating investment risk tolerance as the only measure of financial risk.

### 7. Communication Preferences

Email is the most preferred communication channel, selected by approximately 30% of clients, followed by App at nearly 26% and Phone at around 20.8%. In-person communication represents approximately 16.2%, while Chat is the least preferred at approximately 6.52%.

Quarterly check-ins are the most common engagement frequency, followed by Biannual and Monthly check-ins. Ad-hoc check-ins are the least common.

### 8. Client Satisfaction

Overall client satisfaction is moderate rather than highly positive, with an average score of 2.72 out of 5.

Communication channel alone has little effect on satisfaction, with scores ranging from 2.69 to 2.76, despite Email and App being the most preferred channels.

However, when communication channel and check-in frequency are considered together, satisfaction varies more noticeably. Chat + Monthly has the highest satisfaction score (3.15), while App + Ad-hoc has the lowest (2.44).

Overall, regular client engagement appears to have a stronger influence on satisfaction than communication channel preference alone.


### 9. Risk, Employment and Satisfaction

The relationship between risk category and satisfaction is not simply linear.

Average satisfaction across combinations of risk category and employment status ranges from approximately 1.99 to 3.21. Unemployed High Risk clients report the highest satisfaction at 3.21, while Unemployed Low Risk clients report the lowest at 1.99. Self-employed clients show relatively consistent satisfaction across Low and Moderate Risk categories.

This suggests that employment status and client circumstances may interact with risk category when explaining satisfaction, rather than risk category alone determining satisfaction.

### 10. Debt-to-Income and Portfolio-to-Income Risk

The analysis identifies important financially vulnerable subgroups.

Unemployed High Risk clients have an exceptionally high portfolio-to-income ratio of approximately 36.9, suggesting that their portfolio value is very large relative to their reported income. This may represent accumulated or pre-existing wealth rather than wealth generated from current earnings.

At the other extreme, Unemployed Low Risk clients have the highest debt-to-income ratio of approximately 5.92, indicating substantial financial pressure relative to income.

These two ratios therefore provide complementary perspectives: portfolio-to-income identifies unusual wealth relative to current income, while debt-to-income highlights financial strain.

### 11. Financial Goals

Retirement is the dominant financial goal, with 1,481 clients selecting it, followed by Healthcare with 542 clients. Business Investment is the least common goal with 188 clients.

The target-amount analysis also shows that retirement goals occupy the highest values, with particularly large targets among some marital-status groups. Estate Planning forms another high-value goal category. The analysis indicates that the type of financial goal has a stronger influence on target amount than marital status.

### 12. Further Supporting Analysis(Pairplot and Correlation Analysis)

The pairplot provides an overall view of relationships between age, income, debt, portfolio value and credit score across risk categories.

The strongest patterns are seen among the financial variables rather than age. High Risk clients generally appear at lower levels of income, credit score and investment portfolio, while Low Risk clients generally appear at higher levels. Moderate Risk clients form the largest and most widely distributed group.

The correlation matrix provides several particularly important relationships:

- Annual Income ↔ Monthly Expenses: r = 0.89
- Annual Income ↔ Credit Score: r = 0.69
- Annual Income ↔ Total Debt: r = 0.66
- Age ↔ Investment Portfolio: r = 0.51
- Session Duration ↔ User Satisfaction: r = 0.48

These results indicate strong associations between income and other financial variables, as well as a meaningful relationship between session duration and satisfaction. Importantly, correlation indicates association rather than causation.
